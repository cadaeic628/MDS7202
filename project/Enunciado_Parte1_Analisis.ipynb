```python
# Tu código aquí...
# Recuerda que puedes usar múltiples celdas.
# Pregunta 1: Información general del dataframe y Estadísticos descriptivos de las variables numéricas
df_tickets.info()
df_tickets.describe()
```

```python
# Pregunta 2: Distribución de la variable objetivo
target_col = "Nivel_Prioridad"

conteo = df_tickets[target_col].value_counts()
porcentaje = conteo / conteo.sum() * 100

resumen_objetivo = pd.DataFrame({"Cantidad": conteo, "Porcentaje": porcentaje})

print(resumen_objetivo)

fig = px.histogram(df_tickets, x=target_col)

fig.show()
```

```python
# Pregunta 3: Distribución de características numéricas y categóricas

num_cols = ["N_Caracteres_Ticket", "Tiempo_de_Resolución_Horas", "Usuario-Antiguedad_Cuenta_Dias"]

cat_cols = ["Categoría_Problema", "Canal_Ticket", "Usuario-Tipo_de_Cuenta"]

fig = make_subplots(rows=2, cols=3, subplot_titles=num_cols + cat_cols)

for i, col in enumerate(num_cols):
    hist = px.histogram(df_tickets, x=col)
    fig.add_trace(hist.data[0], row=1, col=i + 1)

for i, col in enumerate(cat_cols):
    hist = px.histogram(df_tickets, x=col)
    fig.add_trace(hist.data[0], row=2, col=i + 1)

fig.show()
```

```python
# Pregunta 4: Correlación entre variables numéricas

corr = df_tickets[num_cols].corr()

fig = px.imshow(corr, text_auto=True)

fig.show()
```

```python
# Pregunta 5: Bins, contingencias y subplots

df_tickets["Bin_N_Caracteres_Ticket"] = pd.qcut(
    df_tickets["N_Caracteres_Ticket"], q=3, labels=["Bajo", "Medio", "Alto"]
)

df_tickets["Bin_Tiempo_de_Resolución_Horas"] = pd.qcut(
    df_tickets["Tiempo_de_Resolución_Horas"], q=3, labels=["Bajo", "Medio", "Alto"]
)

df_tickets["Bin_Usuario_Antiguedad_Cuenta_Dias"] = pd.qcut(
    df_tickets["Usuario-Antiguedad_Cuenta_Dias"], q=3, labels=["Bajo", "Medio", "Alto"]
)

variables = [
    "Categoría_Problema",
    "Canal_Ticket",
    "Usuario-Tipo_de_Cuenta",
    "Bin_N_Caracteres_Ticket",
    "Bin_Tiempo_de_Resolución_Horas",
    "Bin_Usuario_Antiguedad_Cuenta_Dias",
]

fig = make_subplots(rows=2, cols=3, subplot_titles=variables)

for i, col in enumerate(variables):
    tabla = pd.crosstab(df_tickets[col], df_tickets["Nivel_Prioridad"], normalize="index")

    heatmap = px.imshow(tabla, text_auto=True)

    fig.add_trace(heatmap.data[0], row=i // 3 + 1, col=i % 3 + 1)

fig.show()
```

```python
# Pregunta 6: Reclamos por mes y nivel de prioridad

df_tickets["Fecha_Envío"] = pd.to_datetime(df_tickets["Fecha_Envío"])

df_tickets["Fecha_Envío_Mes"] = df_tickets["Fecha_Envío"].dt.to_period("M").astype(str)

reclamos_mes = df_tickets.groupby(["Fecha_Envío_Mes", "Nivel_Prioridad"]).size().reset_index(name="Cantidad")

fig = px.line(reclamos_mes, x="Fecha_Envío_Mes", y="Cantidad", color="Nivel_Prioridad")

fig.show()
```
